```python
from pathlib import Path
import pandas as pd
import numpy as np

# 1. Folder containing newly generated regression results
MY_RESULTS_DIR = Path(
    "/content/drive/MyDrive/ecohydro_project/Project 2/Stage 1/Benchmark_Reproduction/GPCC_GPCP_Results"
)

# 2. Folder containing the authors' original GitHub results
AUTHOR_RESULTS_DIR = Path(
    "/content/drive/MyDrive/ecohydro_project/Project 2/Stage 1/GitHub_Repository/reg_results"
)

# 3. Results to compare
files_to_compare = [
    "GPCP_regression_results.csv",
    "GPCC_regression_results.csv"
]

# 4. Numerical columns to compare
columns_to_compare = [
    "coef_P", "se_P", "pval_P",
    "coef_T", "se_T", "pval_T",
    "coef_Gp", "se_Gp", "pval_Gp",
    "coef_GpP", "se_GpP", "pval_GpP",
    "R2", "model_pval", "vcov"
]

all_reports = []

for filename in files_to_compare:

    my_file = MY_RESULTS_DIR / filename
    author_file = AUTHOR_RESULTS_DIR / filename

    print("\n" + "=" * 75)
    print("COMPARING:", filename)
    print("=" * 75)

    if not my_file.exists():
        print("ERROR: My generated file is missing.")
        continue

    if not author_file.exists():
        print("ERROR: Author's reference file is missing.")
        continue

    my_df = pd.read_csv(my_file)
    author_df = pd.read_csv(author_file)

    if len(my_df) != 1 or len(author_df) != 1:
        print("ERROR: Expected exactly one result row in each file.")
        continue

    my_row = my_df.iloc[0]
    author_row = author_df.iloc[0]

    report = []

    for col in columns_to_compare:

        if col not in my_df.columns or col not in author_df.columns:
            print(f"Missing column: {col}")
            continue

        my_value = float(my_row[col])
        author_value = float(author_row[col])

        # Compare values rounded to three significant figures
        my_3sf = format(my_value, ".3g")
        author_3sf = format(author_value, ".3g")

        match = my_3sf == author_3sf

        report.append({
            "File": filename,
            "Parameter": col,
            "My result": my_value,
            "Author result": author_value,
            "My result (3 sf)": my_3sf,
            "Author result (3 sf)": author_3sf,
            "Absolute difference": abs(my_value - author_value),
            "Match to 3 sf": match
        })

    report_df = pd.DataFrame(report)
    all_reports.append(report_df)

    display(report_df)

    if not report_df.empty:
        if report_df["Match to 3 sf"].all():
            print("\nRESULT: ALL CHECKED VALUES MATCH TO THREE SIGNIFICANT FIGURES.")
        else:
            print("\nRESULT: SOME VALUES DO NOT MATCH TO THREE SIGNIFICANT FIGURES.")

# 5. Combine and save the comparison report
if all_reports:
    final_report = pd.concat(all_reports, ignore_index=True)

    report_path = MY_RESULTS_DIR / "Stage_1_2_GPCP_GPCC_comparison.csv"
    final_report.to_csv(report_path, index=False)

    print("\n" + "=" * 75)
    print("COMPARISON SUMMARY")
    print("=" * 75)

    print("Total parameters checked:", len(final_report))
    print("Matching parameters:", final_report["Match to 3 sf"].sum())
    print("Non-matching parameters:", (~final_report["Match to 3 sf"]).sum())
    print("Overall match:", final_report["Match to 3 sf"].all())
    print("\nComparison report saved at:")
    print(report_path)
```
